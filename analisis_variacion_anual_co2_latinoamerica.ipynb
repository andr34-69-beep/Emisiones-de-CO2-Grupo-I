# Variación porcentual anual de las emisiones de CO₂ en países latinoamericanos

Este análisis utiliza los datos de **Our World in Data** para estudiar cuánto aumentaron o disminuyeron las emisiones de CO₂ respecto al año anterior.

**Fuente:** [Annual change in CO₂ emissions, OWID](https://ourworldindata.org/grapher/change-co2-annual-pct)

**Licencia y atribución:** los datos pertenecen a Our World in Data y a sus fuentes originales.

## 1. Importación y preparación de los datos

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option("display.float_format", lambda value: f"{value:.2f}")
sns.set_theme(style="whitegrid")

ruta = "change-co2-annual-pct.csv"
datos = pd.read_csv(ruta)
columna_variacion = datos.columns[-1]

print(datos.columns.tolist())
datos.head()
```

```python
paises = [
    "Argentina", "Bolivia", "Brazil", "Chile", "Colombia",
    "Costa Rica", "Ecuador", "Mexico", "Peru", "Uruguay"
]

latinoamerica = datos[
    datos["Entity"].isin(paises) & datos["Year"].ge(1990)
].copy()
latinoamerica = latinoamerica.rename(columns={columna_variacion: "Variacion_pct"})

print(f"Países encontrados: {latinoamerica['Entity'].nunique()}")
print(sorted(latinoamerica["Entity"].unique()))
print(f"Período: {latinoamerica['Year'].min()}–{latinoamerica['Year'].max()}")
```

## 2. Evolución desde 1990

Un valor positivo indica que las emisiones aumentaron respecto al año anterior; uno negativo indica una reducción.

```python
plt.figure(figsize=(14, 7))
sns.lineplot(data=latinoamerica, x="Year", y="Variacion_pct", hue="Entity", linewidth=1.8)
plt.axhline(0, color="black", linewidth=1)
plt.title("Variación porcentual anual de las emisiones de CO₂: Latinoamérica", fontsize=15)
plt.xlabel("Año")
plt.ylabel("Cambio anual (%)")
plt.legend(title="País", ncol=2)
plt.tight_layout()
plt.show()
```

## 3. Comparación entre países

La siguiente tabla muestra el promedio anual y el rango observado en el período seleccionado.

```python
resumen = latinoamerica.groupby("Entity")["Variacion_pct"].agg(
    Promedio="mean", Máximo="max", Mínimo="min"
).round(2).sort_values("Promedio", ascending=False)
resumen
```

```python
plt.figure(figsize=(12, 6))
sns.barplot(data=resumen.reset_index(), x="Promedio", y="Entity", hue="Entity", palette="viridis", legend=False)
plt.axvline(0, color="black", linewidth=1)
plt.title("Promedio de la variación anual desde 1990")
plt.xlabel("Cambio promedio (%)")
plt.ylabel("País")
plt.tight_layout()
plt.show()
```

## 4. Colombia

Los datos de Colombia se separan para observar su tendencia reciente y sus mayores cambios interanuales.

```python
colombia = latinoamerica[latinoamerica["Entity"] == "Colombia"].sort_values("Year")
colombia.tail(12)
```

```python
plt.figure(figsize=(12, 5))
plt.plot(colombia["Year"], colombia["Variacion_pct"], marker="o", linewidth=2, color="#0072B2")
plt.axhline(0, color="black", linewidth=1)
plt.title("Variación anual de las emisiones de CO₂ en Colombia")
plt.xlabel("Año")
plt.ylabel("Cambio anual (%)")
plt.tight_layout()
plt.show()

print("Último dato disponible:")
print(colombia.iloc[-1][["Year", "Variacion_pct"]].to_string())
```

## 5. Interpretación

- La variación positiva significa aumento respecto al año anterior.
- La variación negativa significa reducción respecto al año anterior.
- Las crisis económicas, la evolución de la producción energética y eventos extraordinarios pueden explicar cambios abruptos.
- Los promedios describen tendencias generales y no sustituyen el análisis año por año.
